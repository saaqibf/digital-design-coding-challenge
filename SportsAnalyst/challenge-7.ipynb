```python
#❗️Run, then change everything about your creation
import pandas as pd
df = pd.read_csv("data.csv")
my_creation = df.iloc[0].copy()
my_creation["Player"] = "YOUR NAME HERE"
# change any numbers, for example:
# my_creation["PointsPerGame"] = 99
# my_creation["SalaryMillions"] = 1
df2 = pd.concat([df, my_creation.to_frame().T], ignore_index=True)
df2.tail(3)
```

```python
#❗️Run: where does your creation land on the value chart?
import matplotlib.pyplot as plt
df2_num = df2.copy()
for col in ["PointsPerGame", "SalaryMillions"]:
    df2_num[col] = pd.to_numeric(df2_num[col])
plt.figure(figsize=(9,5))
plt.scatter(df2_num["SalaryMillions"], df2_num["PointsPerGame"])
for _, row in df2_num.iterrows():
    plt.annotate(str(row["Player"]).split()[0], (row["SalaryMillions"], row["PointsPerGame"]), fontsize=7)
plt.xlabel("SalaryMillions")
plt.ylabel("PointsPerGame")
plt.title("My world")
plt.show()
```
