```python
#❗️Run: the mystery of shift 13
import byte as b
once = b.encode("WHY IS THIRTEEN SPECIAL", 13)
twice = b.encode(once, 13)
print("encoded once: ", once)
print("encoded twice:", twice)
```

Noticed what shift 13 does when you apply it twice? There are 26 letters. Half of 26 is 13. Encoding twice slides letters all the way around the circle, back to the start. Programmers call this cipher ROT13 and it's a real thing on the internet.
